### Iterables vs Iterators 🔥

```python
# 1. Iterable
# An iterable is an object you can loop over.
```

```python
lst = [1,2,3,4]
text = "hello"
tup = (1,2,3)
```

```python
for x in lst:
    print(x)

# we can iteravte over list
# So list is an iterable.
```

```python
# 2. Iterator
# An iterator is an object that gives values one at a time.
```

```python
# You use:
iterator = (x for x in [1,2,3])
next(iterator)
```

```python
lst = [10, 20, 30]

it = iter(lst)

print(next(it))
print(next(it))
print(next(it))
```

```python
# lst itself is an iterable.
# it is an iterator created from that iterable.
```

```python
# Connection with for loop
# When you write:

for x in lst:
    print(x)

# Python roughly does this internally:

# it = iter(lst)

# while True:
#     try:
#         x = next(it)
#         print(x)
#     except StopIteration:
#         break

it = iter(lst)
while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
```

```python
# So a for loop automatically:

# Gets an iterator using iter()
# Calls next()
# Calls next() again
# Stops when StopIteration occurs

# You normally don't have to write this manually.
```

```python
# Quick Difference
# Iterable → Something you can get values from
# Iterator → Something that gives the next value
```

```python
# A generator is both iterable and an iterator.

gen = (x * 2 for x in [1, 2, 3])

# You can directly do:

print(next(gen))

# because a generator is already an iterator.

# You can also loop over it:

for x in gen:
    print(x)
```

```python
# Unlike a generator, a list can be iterated again because each for loop creates a new iterator internally.
```

```python
# Zip() is also an iterator
names = ["Shri", "Rahul", "Aman"]
marks = [90, 85, 95]

z = zip(names, marks)

# z is a zip object, which behaves as an iterator.
```

```python
# enumerate() is also an Iterator
names = ["A", "B", "C"]

e = enumerate(names, start=1)

print(next(e))
print(next(e))
print(list(e))
    
```

```python
# range is an iterable, not an iterator
r = range(1,4)

for x in r:
    print(x)
    
# next(r) will give error
# We need:
it = iter(r)

print(next(it))
```

```python
# iter() on an Iterator 🔥

# This is an important rule:
iter(iterator)
# usually returns the same iterator.
```

```python
lst = [10,20,30,40]
itr1 = iter(lst)
itr2 = iter(itr1)

print(next(itr1))
print(next(itr1))
print(next(itr2))
print(next(itr2))
print(itr1 is itr2)
```

```python
# Creating your own iterator
# For an object to behave as an iterator, it needs:

# __iter__()
# __next__()
```

```python
class count:
    def __init__(self,n):
        self.n = n
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.n>3:
            raise StopIteration
        
        value = self.n
        self.n += 1
        return value
```

```python
c = count(1)
print(next(c))
print(next(c))
print(next(c))
```
