### Decorator

```python
# A decorator is a function that takes another function, adds some extra behavior to it, and returns a new function.
# The simplest structure is:

# def decorator(func):

    # def wrapper():
        # Extra work before
        # func()
        # Extra work after

    # return wrapper
```

```python
def decorator(func):
    
    def wrapper():
        print("Before")
        
        func()
        
        print("After")
        
    return wrapper
```

```python
def greet():
    print("Hello")
```

```python
greet = decorator(greet)
```

```python
# Step 1: Pass greet into decorator
# decorator(greet)

# So inside:
# func = greet
```

```python
# Step 2: decorator() creates wrapper
# def wrapper():
#     print("Before")
#     func()
#     print("After")

# Remember:
# func → original greet
```

```python
# Step 3: Return wrapper
# return wrapper
```

```python
# So the structure becomes:
# greet
#   ↓
# wrapper()
#   ↓
# "Before"
#   ↓
# original greet()
#   ↓
# "Hello"
#   ↓
# "After"
```

```python
# The @decorator Syntax 🔥

# Python gives us a shorter way to write:

def hello():
    print("Hello")

hello = decorator(hello)

# Instead, we write:
@decorator
def hello():
    print("Hello")

# These two are essentially equivalent:
# Take this function
# ↓
# Pass it to decorator
# ↓
# Replace the function with what decorator returns
```

```python
def A(func):
    def wrapper():
        print("A1")
        func()
        print("A2")
    return wrapper

def B(func):
    def wrapper():
        print("B1")
        func()
        print("B2")
    return wrapper

@A
@B
def C():
    return "Hello"

print(C())
```

```python
# For:

# @A
# @B
# def func():

# Python does:

# func = A(B(func))

# Execution goes:

# A Before
# B Before
# Original function
# B After
# A After
```

```python
def dec1(func):
    def wrapper():
        print("Dec1 Before")
        func()
        print("Dec1 After")
    return wrapper

def dec2(func):
    def wrapper():
        print("Dec2 Before")
        func()
        print("Dec2 After")
    return wrapper
```

```python
@dec1
@dec2
def greet():
    print("Hello")
```

```python
# This is equivalent to greet = dec1(dec2(greet))
greet()
```

```python
def A(func):
    def wrapper():
        print("1")
        result = func()
        print("4")
        return result
    return wrapper

def B(func):
    def wrapper():
        print("2")
        result = func()
        print("3")
        return result
    return wrapper

@A
@B
def hello():
    print("Hello")
    return 100

print(hello())
```

```python
# functools.wraps 🔥
# When we decorate a function, the original function gets replaced by wrapper.
```

```python
def decorator(func):
    def wrapper():
        return func()
    return wrapper

@decorator
def greet():
    return "Hello"

print(greet.__name__)
```

```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper():
        return func() + "!"
    return wrapper


@decorator
def greet():
    return "Hello"


print(greet())
print(greet.__name__)
```

```python
# print(greet.__name__)

# means:
# What is the name of the function that greet currently refers to?
```

```python
# Let's first convert:

# @wraps(func)
# def wrapper():

# into the same decorator logic you already learned:

# def wrapper():
#     return func()

# wrapper = wraps(func)(wrapper)

# So wraps(func) creates something that updates wrapper's metadata using the metadata of func.
```

```python
# Remember:

# func → original greet

# Original greet has:

# func.__name__ = "greet"

# Wrapper originally has:

# wrapper.__name__ = "wrapper"

# After:

# @wraps(func)

# its metadata is updated roughly so that:

# wrapper.__name__ = func.__name__

# Therefore:

# wrapper.__name__

# becomes:
#     "greet"

# The actual function is still the wrapper.
```

```python
# wraps does not remove the wrapper.
# The wrapper still executes.
# It just makes the wrapper preserve information about the original function.
```

```python
# check odd even

def check_odd_even(func):
    
    def wrapper(*args,**kwargs):
        if args[0]%2==0:
            return func()
        return "Odd"
    return wrapper

@check_odd_even
def check():
    return "Even" 

print(check(2))
print(check(3))
print(check(4))
```

```python
# compute squares
def calclualte(func):
    
    def wrapper(*args,**kwargs):
        print("Calculate")
        result = func(*args,**kwargs)
        return result
    return wrapper

@calclualte
def square(a):
    return a*a 

@calclualte
def cube(a):
    return a*a*a 

print(square(4))
print(cube(3))
```

```python
def inner():
    yield 10
    yield 20


def outer():
    yield 5
    yield from inner()
    yield 30


print(list(outer()))
```

```python
# | List                            | Generator                        |
# | ------------------------------- | -------------------------------- |
# |  []                             | ()                               |
# | Stores all values               | Produces values one by one       |
# | Can iterate multiple times      | Usually consumed after iteration |
# | Uses more memory for large data | Usually more memory-efficient    |
# | Values available immediately    | Values generated when needed     |
```

```python
# Generators: Phase Summary 🧠

# You now know:

# yield                  → give a value and pause
# next(generator)        → resume until next yield
# generator exhaustion   → StopIteration
# for loop               → automatically handles StopIteration
# yield vs return        → pause vs permanently end
# generator expressions  → (...) instead of [...]
# generators are consumed
# yield from             → yield values from another iterable/generator
# lazy execution         → generate values only when needed
# memory efficiency      → don't store all values at once
```

```python
class count:
    def __init__(self,n):
        self.count = 0
        self.n = n
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        
        self.count += 1
        return self.count * 10
    
    
class Number:
    def __init__(self,n):
        self.n = n
        
    def __iter__(self):
        return count(self.n)
    
nums = Number(2)

print(list(nums))
print(list(nums))
```
