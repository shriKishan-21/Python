```python
def check(num):
    if num%2==0:
        raise ValueError("Even number")
    
    return "Odd Number"

try:
    print(check(11))
except ValueError as e:
    print(e)
    
finally:
    print("Thankyou")
    
```

```python
class EvenOddException(Exception):
    pass

def check(num):
    if num%2==0:
        raise EvenOddException("Even number")
    
    return "Odd Number"

try:
    print(check(11))
except EvenOddException as e:
    print(e)
    
finally:
    print("Thankyou")
    
```

```python
def check(x):
    if x == 0:
        raise ValueError("Zero not allowed")
    return 100 / x


try:
    print(check(0))
except ValueError as e:
    print(e)
finally:
    print("Finished")
```

```python
# Re-raising an Exception with raise

# There's one more important raise concept.

# Sometimes you catch an exception, do something with it, but then want to send the same exception upward again.
```

```python
# Example:
try:
    x = 10 / 0

except ZeroDivisionError:
    print("Error happened")
    raise
```

```python
# Here:
# raise
# without an exception name means:
# Raise the currently handled exception again.
```

```python
def divide(a,b):
    try:
        return a/b
    
    except ZeroDivisionError:
        print("Logging error")
        raise
    
    
try:
    divide(10,0)
    
except ZeroDivisionError:
    print("Handled Outside")
```
