```python
# Set
```

```python
# Its main properties:

# 1. Stores unique values
# 2. No duplicates
# 3. Mutable
# 4. Not accessed using indexes
# 5. Fast membership checking
```

```python
s = {1,2,3,4,5,6}
print(s)
```

```python
s = {1,2,2,2,3,4}
print(s)
# it automatically remove duplicates
```

```python
s[0]
# indexing is not allowed
```

```python
s.add(10)
print(s)
```

```python
s.remove(10)
print(s)
```

```python
if 10 in s:
    print("Find")
```

```python
print(3 in s)
```

```python
lst = [10, 20, 30, 40]
30 in lst
```

```python
# Usually requires checking elements one by one → O(n).
```

```python
# But in set
s = {10, 20, 30, 40}
30 in s
```

```python
# Average-case membership lookup is O(1), which makes sets extremely useful in DSA.
```

```python
s = {1,2,3,4}
s.remove(2)
print(s)
```

```python
s.remove(10)
# Raise error because 10 doesnt exist
```

```python
s.discard(10)
# discard doesn't give nothing even 10 doesn't exist
```

```python
s = {1,2,3}
s.update([4,5,6])
print(s)
```

```python
a = {1,2,3,4}
b = {3,4,5,6}
```

```python
# Union
c = a|b
print(c)
```

```python
# Intersection
d = a&b
print(d)
```

```python
# Difference
e = a-b
print(e)
f = b-a
print(f)
# Difference is directional.
```

```python
# Symmetric Difference
g = a^b
print(g)
```
