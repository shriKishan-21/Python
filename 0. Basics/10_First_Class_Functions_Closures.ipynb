```python
# Functions are objects in Python
# You can assign a function to a variable:

def greet():
    return "Hello"

x = greet

print(x())
# Notice:
# x = greet
# We are not calling greet. We are storing the function itself in x.
# But:
# x = greet()
# would call the function and store its returned value.
```

### First Class Functions

```python
# Functions Can Be Passed as Arguments 🔥
# Since functions are objects, you can pass a function to another function.
def square(x):
    return x*x

def apply(func,value):
    return func(value)

print(apply(square,5))
```

```python
def divide(x):
    return x//2

def math(func,value):
    return func(value)*value

print(math(divide,6))
```

### Functions Can Return Functions 🔥

```python
# Just as Python can:
# store a function in a variable
# pass a function as an argument
# it can also return a function.
```

```python
def outer():
    def inner():
        print("Hello")
        
    return inner

x = outer()
print(x())
```

### Returning Different Functions 🔥

```python
def choose_operation(choice):
    def add(x):
        return x+10
    def multiply(x):
        return x*10
    if choice == "add":
        return add
    else:
        return multiply
        
```

```python
func = choose_operation("multiply")
```

```python
print(func(5))
```

### Closures 🔥

```python
# Now we combine what you just learned:

# Functions can be inside functions
# Functions can be returned
# Functions can access variables from their outer function
```

```python
def outer(x):
    def inner():
        return x*2
    
    return inner
```

```python
func = outer(10)
print(func())
```

```python
# A closure is a function that remembers and can use variables from its enclosing scope, even after the outer function has finished.
```

```python
# local, nonlocal, global
```

```python
# local     → current function
# nonlocal  → enclosing function
# global    → global scope
```

```python
x = 100

def outer():
    y=10
    
    def inner():
        z=5  # local
        nonlocal y # enclosing variable
        global x # gloabal variable
        
```

```python
# A closure can remember more than one variable from its enclosing function.
```

```python
def create_operations(a,b):
    def calculate(x):
        return x*a+b
    return calculate

func = create_operations(2,5)
print(func(10))
```

```python
# Closure with a List
```

```python
def create_list():
    items = []
    
    def add_items(x):
        items.append(x)
        return items
    
    return add_items

add = create_list()

print(add(10))
print(add(20))
print(add(30))
```

```python
# A closure can remember:

# Numbers
# Strings
# Lists
# Dictionaries
# Other objects

# And if the object is mutable, you can modify that same remembered object across multiple calls.
```

```python
# Next: Why Closures Are Useful?
# Closures are useful when you want a function to remember some configuration or state.
```

```python
def create_generator(name):
    def greet():
        return f"Hello, {name}"
    
    return greet
```

```python
var = create_generator("ashu")
print(var())
```

```python
def outer():
    x = 10

    def inner():
        return x + 5

    return inner

# Here, inner() only reads x, so no nonlocal is needed.
```

```python
# But if you want to reassign x:

def outer():
    x = 10

    def inner():
        nonlocal x
        x += 5
        return x

    return inner

# Now nonlocal x tells Python:
# Don't create a new local x. Use the x from outer().
```

```python
def power(exponent):
    def calculate(number):
        return number**exponent
    return calculate
    
func2 = power(2)
func3 = power(3)
print(func2(4))
print(func3(4))
```

```python
# Let's summarize what a closure is:

def outer(x):
    def inner(y):
        return x + y
    return inner

# When you do:

f = outer(10)

# f is a function, but it also remembers x = 10.
# You can think of it as:

# f
# ├── function: inner(y)
# └── remembered environment:
    #    x = 10
```
