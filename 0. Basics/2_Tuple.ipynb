```python
# Tuples have very few methods because they are immutable.

# The main ones are:

# count()
# index()
```

```python
tup = (10,20,30,40,50,20,20)

print(tup.count(20))
print(tup.index(30))
```

```python
# count(x) → how many times x appears
# index(x) → index of the first occurrence
# index(x,y) -> index of x but the start index is y
```
