```python
# [] vs .get()
d = {
    'name':'shri',
    'age':'21'
}
```

```python
print(d['name'])
print(d['college'])
# since college entry doesn't exist it will raise error
```

```python
print(d.get('name'))
print(d.get('college'))
#  .get will not raise an error
```

```python
# d[key]              → Error if key doesn't exist
# d.get(key)          → None if key doesn't exist
# d.get(key, default) → Default value if key doesn't exist
```

```python
d = {
    "name": "Shri",
    "age": 21,
    "branch": "Data Science"
}
```

```python
for key in d.keys():
    print(key)
```

```python
for value in d.values():
    print(value)
```

```python
for key,value in d.items():
    print(key,"->",value)
```

```python
lst = [2,2,3,1,2,4]
freq = {}
for x in lst:
    if x in freq:
        freq[x] += 1
    else:
        freq[x]=1
        
print(freq)
```

```python
freq = {}
for x in lst:
    freq[x] = freq.get(x,0)+1

print(freq)
```

```python
# setdefault() vs .get()
d = {}
```

```python
print(d.get('a',0))
# But it does not add "a" to the dictionary.
print(d)
```

```python
print(d.setdefault('a',0))
print(d)
```

```python
d = {'a':10}
d.setdefault('a',0)
print(d)
# It does not replace 10 with 0.
```

```python
nums = [1,2,3,4]
squares = {}
for x in nums:
    squares[x] = x*x
    
print(squares)

# one line for this
square = {x:x*x for x in nums}
print(square)
```

```python
# if
sq1 = {x:x**2 for x in nums if x%2==0}
print(sq1)

# if-else
sq2 = {x:"Even" if x%2 ==0 else "odd"
       for x in nums}
print(sq2)
```

```python
# Dictionary transformations
student = {
    "name": "Shri",
    "age": 21,
    "branch": "Data Science"
}
```

```python
# 1. Swap keys and values
swapped = {value:key for key,value in student.items()}
print(swapped)
```

```python
# Values must be unique if you want to safely swap them. Otherwise, duplicate values overwrite earlier keys.
```

```python
# zip()
names = ["shri", "Rahul", "Aman"]
marks = [90, 85, 95]

for name,mark in zip(names,marks):
    print(name,mark)
```

```python
# Important: Different lengths
a = [1, 2, 3, 4]
b = ["a", "b"]

c = list(zip(a,b))
print(c)
# zip() stops when the shortest iterable ends.
```

```python
names = ["Shri", "Rahul", "Aman"]
marks = [90, 85, 95]

d = dict(zip(names,marks))

print(d)
```

```python
# enumerate
lst = [10,20,30,40,50]

for i,val in enumerate(lst,start=3):
    print(i,val)
```

```python
d = dict(enumerate(lst))
print(d)
```

```python
# enumerate() + zip()
names = ["Shri", "Rahul", "Aman"]
marks = [90, 85, 95]

for i,(name,mark) in enumerate(zip(names,marks),start=1):
    print(i,name,mark)
```
