```python
def square(x):
    print(x*x)
    
square(2)
```

```python
# Lambda function
sq = lambda x : x*x
sq(2)
```

```python
add = lambda a,b:a+b
add(2,3)
```

```python
# General syntax
# lambda arguments: expression
# The expression is automatically returned.
```

```python
check = lambda x : "Even" if x%2==0 else "Odd"
check(3)
```

```python
# Where Lambda Becomes Really Useful — sorted()
# Suppose we have:
students = [
    ("Shri", 85),
    ("Rahul", 95),
    ("Aman", 75)
]

# If we do:
print(sorted(students))

# Python sorts by the first element of each tuple (name).
# But what if we want to sort by marks?

sorted_students = sorted(students,key=lambda x:x[1])

print(sorted_students)

# Here:
# lambda x: x[1]
# means:
# For each tuple, use its second value as the sorting key.
```

```python
# Descending sorting
# use: reverse=True
result = sorted(students,key=lambda x : x[1],reverse=True)
print(result)
```

```python
words = ["cat", "elephant", "dog", "ant"]

result = sorted(words,key=lambda x : (len(x),x))
print(result)
```

### map()

```python
# map() applies a function to every element of an iterable.
```

```python
nums = [1,2,3,4]
result = []
for x in nums:
    result.append(x*x)
    
print(result)
```

```python
result = list(map(lambda x : x*x,nums))
result
```

```python
# Structure
# map(function, iterable)
```

### filter()

```python
# filter() is similar to map(), but instead of transforming every element, it keeps only elements that satisfy a condition.
```

```python
nums = [1,2,3,4,5]
result = []
for x in nums:
    if x%2==0:
        result.append(x)
        
print(result)
```

```python
result = list(filter(lambda x:x%2==0,nums))
result
```

```python
# Structure
# filter(condition_function, iterable)
```

```python
# The function should return:
# True  → keep the element
# False → discard the element
```

```python
nums = [1,2,3,4,5,6]
result = list(filter(lambda x : (x**2)%2==0,nums))
result 
```

```python
# map() vs filter()
```

```python
# map()
# Transforms every element
nums = [1, 2, 3]
list(map(lambda x: x * 2, nums))
```

```python
# filter()
# Keeps selected elements
nums = [1,2,3]
list(filter(lambda x:x>1,nums))
```

```python
nums = [1,2,3,4,5,6]
result = list(map(lambda x : x*x,filter(lambda x : x%2==0,nums)))
result
```

### any() and all() 🔥

```python
# These are small but very useful Python functions.
```

```python
# any()
# Returns True if at least one element is truthy.
nums = [False,False,True,False]
print(any(nums))
```

```python
nums = [1,2,3,4]
print(any(x%2==0 for x in nums))
```

```python
nums = [1,3,5,7]
print(any(x%2==0 for x in nums))
```

```python
# all()
# Returns True only if every element is truthy.
nums = [False,False,True,False]
print(all(nums))
```

```python
nums = [2,4,6,8]
print(all(x%2==0 for x in nums))
```

```python
nums = [2,4,6,9]
print(all(x%2==0 for x in nums))
```
