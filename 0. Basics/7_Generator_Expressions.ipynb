### Generator Expressions 🔥

```python
nums = [1,2,3,4]
any(x > 6 for x in nums)
# This part:
# x > 6 for x in nums
# is a generator expression.
```

```python
# list comprehension
nums = [1,2,3,4]
result = [x*x for x in nums]
result
# This immediately creates a list with the tranformation

# generator expression
gen = (x*x for x in nums)
gen
# It does not immediately create and store the complete list of results.
```

```python
# Why useful?
# For large datasets, generators can be more memory-efficient because values are produced one at a time instead of storing everything at once.
```

```python
data = [1,2,3,4,5,6]
gen = (x*x for x in data)
result = sum(x*x for x in data)
result
```

### List Comprehension

```python
nums = [1,2,3,4]
result = [x*x for x in nums]
result
```

```python
result = [x**2 for x in nums if x*2>4]
result
```

```python
result = [x**2 if x*2>4 else x for x in nums]
result
```

### Nested List Comprehension

```python
result = []

for i in [1,2,3]:
    for j in [10,20]:
        result.append(i+j)
        
result
```

```python
# nested list comprehension
result = [i+j for i in [1,2,3] for j in [10,20]]
result
```

```python
# [expression
#  for i in outer_iterable
#  for j in inner_iterable]
```

```python
nums = [1,2,3,4,5,6,7,8]
result = []
for x in nums:
    if x>2:
        if x%2==0:
            result.append(x)
            
result
```

```python
result = [x for x in nums if x>2 if x%2==0]
result = [x for x in nums if x>2 and x%2==0]
# both are acceptable
result
```

```python
# Nested List Comprehension with a Condition
result = []
for x in [1,2,3]:
    for y in [1,2,3]:
        if x!=y:
            result.append((x,y))
            
result
```

```python
result = [(x,y) for x in [1,2,3] for y in [1,2,3] if x!=y]
result
```
