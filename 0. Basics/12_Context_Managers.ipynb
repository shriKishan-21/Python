### Context Managers and with

```python
file = open("file.txt","r")
data = file.read()
file.close()
```

```python
data
```

```python
# The problem: if an error happens before:

# file.close()

# the file may not be closed properly.
```

```python
# Using with
with open("file.txt","r") as file:
    data = file.read()
    print(data)
    
```

```python
# After the with block finishes, Python automatically closes the file.
```

```python
class Demo:
    def __enter__(self):
        print("Entering")
        return self
    
    def __exit__(self,exc_type,exc_value,tracback):
        print("Exiting")
```

```python
with Demo() as d:
    print("Inside")
```

```python
class Test:
    def __enter__(self):
        print("Start")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("End")
        return True


with Test():
    print("Inside")
    5 / 0

print("After")
```

```python
# return True from __exit__() does not print True. It tells Python:
# “I handled the exception, so don't raise it further.”
```

```python
# __exit__() with return False

# If __exit__() returns False or None, the exception is not suppressed.

def __exit__(self, exc_type, exc_value, traceback):
    print("End")
    return False
```

```python
class test:
    def __enter__(self):
        print("Enter")
        return self
    
    def __exit__(self,exc_type, exc_value, traceback):
        print("Exit")
        return False
    
with test() as t:
    print("Inside")
    10/0
    
print("After")
```

```python
class Test:
    def __enter__(self):
        print("1. Enter")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("3. Exit")
        return True


with Test() as x:
    print("2. Inside")
    10 / 0

print("4. After")
```
