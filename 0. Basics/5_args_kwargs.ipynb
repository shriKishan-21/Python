```python
# *args
# Suppose you don't know how many arguments a function will receive:
def add(*args):
    print(args)
```

```python
add(10,20,30)
# So args is a tuple.
```

```python
def add(*args):
    total = 0
    for x in args:
        total += x
        
    return total

add(10,20,30)
```

```python
# Important

# args is just a convention. This also works:
def add(*numbers):
    print(numbers)
    
add(10,20)
```

```python
# **kwargs
# This collects keyword arguments into a dictionary:
def student(**kwargs):
    print(kwargs)
    
student(name='shri',age=21)
```

```python
# *args     → collects positional arguments into a tuple
# **kwargs  → collects keyword arguments into a dictionary
```

```python
# But * and ** can also do the opposite: unpack values.
```

```python
nums = [10,20,30]
print(*nums)
```

```python
def add(a,b,c):
    return a+b+c

nums = [10,20,30]
print(add(*nums))
```

```python
def student(name,age):
    print(name,age)
    
d = {
    'name' : 'shri',
    'age' : 21
}

student(**d)
```

```python
# **data
# unpacks the dictionary into keyword arguments:
# student(name="Shri", age=21)
```

```python
# | Syntax                            | Meaning                                   |
# | --------------------------------- | ----------------------------------------- |
# | `*args` in function definition    | Collect positional arguments → tuple      |
# | `**kwargs` in function definition | Collect keyword arguments → dictionary    |
# | `*lst` in function call           | Unpack iterable into positional arguments |
# | `**d` in function call            | Unpack dictionary into keyword arguments  |
```

```python
def demo(a, *args, **kwargs):
    print(a)
    print(args)
    print(kwargs)

demo(10, 20, 30, 40, x=100, y=200)
```

```python
# Default Parameters + Keyword-Only Arguments
def create(name,age=18):
    print(name,age)
    
create('ashu')
```

```python
create('ashu',22)
```

```python
# Keyword-only arguments
def test(a,*,b,c):
    print(a,b,c)
    
test(10,b=20,c=30)
```

```python
test(10,20,30)
# this will not work because b and c are keyword-only.
```

```python
# Positional-Only Arguments
def func(a, b, /, c):
    print(a, b, c)
```

```python
# The / means:
# Everything before / must be passed positionally.
```

```python
func(10, 20, 30)
```

```python
func(10,20,c=30)
```

```python
func(a=10,b=20,c=30)
```

```python
def func(a, /, b, *, c):
    print(a,b,c)
    
func(10,20,c=30)
func(10,b=20,c=30)
```

```python
# a → positional-only
# b → positional OR keyword
# c → keyword-only
```
