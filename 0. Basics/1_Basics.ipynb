```python
names = ["shri","Rahul","Aman"]
marks = [90,85,95]

for idx in range(len(names)):
    print(names[idx],"->",marks[idx])
```

```python
for name,mark in zip(names,marks):
    print(name,"->",mark)
```

```python
# What does zip() do?
# It combines corresponding elements:
#     zip(names, marks)
    
# Conceptually gives:
# ("Shri", 90)
# ("Rahul", 85)
# ("Aman", 95)
```

```python
for idx,(name,mark) in enumerate(zip(names,marks),start=1):
    print(f"Student {idx} : {name}->{mark}")
```

```python
def checkIndex(pattern,arr):
    
    # split the pattern
    parts = pattern.split(",")
    
    # find the positions of starred variable
    start_pos = 0
    
    for i,value in enumerate(parts):
        if value.startswith("*"):
            start_pos = i
            break
    
    # number of normal variable before *
    before_count = start_pos
    
    # number of normal variables after *
    total = len(parts)
    after_count = total - start_pos - 1
    
    # split the array into three parts
    before_values = arr[:before_count]
    
    if after_count == 0:
        start_values = arr[before_count:]
        after_values = []
    else:
        start_values = arr[before_count:-after_count]
        after_values = arr[-after_count:]
    
    print("Before:",before_values)
    print("Start:",start_values)
    print("After:",after_values)
```

```python
pattern = "a,b,c,*d"
arr = [1,2,3,4,5,6]
checkIndex(pattern,arr)
```

```python
a = [10,20,30]
b=a.copy()
b.append(40)
print(a)
print(b)
```

```python
# shallow copy?

# This:
# b = a.copy()

# copies only the outer container:

# a ──► Outer List A
# b ──► Outer List B

# But the objects inside are still shared:

# Outer List A ──► Inner List [1, 2]
#              └──► Inner List [3, 4]


# Outer List B ──► Same Inner List [1, 2]
#              └──► Same Inner List [3, 4]
```

```python
# The key difference 🔥
# Modifying an inner shared object:
# b[0].append(100)

# Changes both a and b.

# Modifying the outer list:
# b.append([5, 6])

# Changes only b.
```

```python
a = [[10,20],30]

b = a.copy()
b.append(40)

c = a.copy()
b[0].append(30)

print(a)
print(b)
print(c)
```
