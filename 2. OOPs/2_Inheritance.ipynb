### Inheritance

```python
# Inheritance means:

# A child class can reuse and extend the behavior of a parent class.
```

```python
class Animal:
    def speak(self):
        print("AnimalSound")
        
class Dog(Animal):
    pass
```

```python
d = Dog()
d.speak()
```

```python
# Method Overriding
```

```python
# Suppose:

class Animal:

    def speak(self):
        print("Animal sound")


class Dog(Animal):

    def speak(self):
        print("Woof")

# Dog has its own speak(), so it overrides the parent's method.
```

```python
d = Dog()
d.speak()
```

```python
# Python searches:

# d
# ↓
# Dog
# ↓
# speak() found ✅
# ↓
# use Dog.speak()

# It doesn't continue to Animal.speak().
```

### super() 🔥

```python
# Sometimes a child class overrides a parent's method but still wants to run the parent's version.
```

```python
class Animal:
    def speak(self):
        print("Animal")
        
class Dog(Animal):
    def speak(self):
        print("Dog")
        super().speak()
```

```python
d = Dog()
d.speak()
```

```python
# super().speak()

# means:
# Go to the parent class and call its speak().

# So:
# Dog.speak()
#    ↓
# print("Dog")
#    ↓
# super().speak()
#    ↓
# Animal.speak()
#    ↓
# print("Animal")
```

```python
# super() with attributes
```

```python
class Animal():
    def __init__(self,name):
        self.name = name
        
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed
```

```python
d = Dog("Bruno","Labrador")
```

```python
# Step-by-step

# Dog.__init__ receives:

# name = "Bruno"
# breed = "Labrador"

# Then:

# super().__init__(name)

# calls:

# Animal.__init__("Bruno")

# which creates:

# d.name → "Bruno"

# Then Dog continues:

# self.breed = breed

# creating:

# d.breed → "Labrador"

# Final object:

# d
# ├── name  → Bruno
# └── breed → Labrador

# So super() lets the child reuse the parent's initialization/behavior instead of duplicating it.
```
