```python
# 4. Methods

# A class can also contain functions.

class Student:

    def introduce(self):
        print("Hello")

# Now:

s1 = Student()
s1.introduce()

# prints:
# Hello

# The function inside a class is called a method.
```

```python
# 5. The confusing part: self

# This is where I want to slow down.

class Student:

    def introduce(self):
        print("Hello")

# When you do:

s1 = Student()
s1.introduce()

# Python internally passes s1 to the method.

# Conceptually:
# Student.introduce(s1)

# Therefore:
# self → s1

# So self means:
# the particular object that is currently using the method.

# For another object:

# s2.introduce()

# Python conceptually does:

# Student.introduce(s2)

# Now:

# self → s2
```

```python
# 🧠 The key idea
# s1.introduce()

# means roughly:

# Student.introduce(s1)

# and:

# s2.introduce()

# means roughly:

# Student.introduce(s2)

# That's why we need self.
```

```python
# 6. Now __init__

# Suppose every student should have a name and age.

# We could do:

class Student:
    pass

s1 = Student()
s1.name = "Shri"
s1.age = 21

# But this is inconvenient.
# Instead, we use:

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
# Now:

s1 = Student("Shri", 21)
s2 = Student("Rahul", 22)
```

```python
# self.name

# This is very important in OOP.

# Look at:
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

# There are actually two different names here:

# self.name = name


# Right side:
# name
# is the parameter.

# Left side:
# self.name

# is an attribute belonging to the object.
```

```python
# So when:

s1 = Student("Shri", 21)

# Python effectively has:
# self → s1
# name → "Shri"
# age  → 21

# Then:
# self.name = name

# becomes:
# s1.name = "Shri"

# And:
# self.age = age

# becomes:
# s1.age = 21
```

```python
# Different objects = different attributes
s1 = Student("Shri", 21)
s2 = Student("Rahul", 22)

# creates:

# s1
# ├── name → Shri
# └── age  → 21

# s2
# ├── name → Rahul
# └── age  → 22

# The class provides the structure, but each object has its own instance data.
```

### Instance Variables vs Class Variables

```python
# Instance variable
class Student:

    def __init__(self, name):
        self.name = name

# Here:

# self.name

# belongs to each individual object.
```

```python
# Class variable

# Now:

class Student:
    school = "ABC School"

    def __init__(self, name):
        self.name = name

# school belongs to the class.
```

```python
# Think:

#              Student class
#                  │
#            school = ABC School
#                  │
#           ┌──────┴──────┐
#           ↓             ↓
#          s1             s2
#        name=Shri      name=Rahul

# Both objects can access:

# print(s1.school)
# print(s2.school)

# Both give:

# ABC School
# ABC School

# But:

# s1.name

# and:

# s2.name

# are different.
```

```python
# 🧠 Very Important Difference
# s1.school = ...

# Changes/creates the attribute for one object:

# s1 → school = XYZ
# s2 → school = ABC (from class)
# Student → school = ABC
# Student.school = ...

# Changes the attribute on the class:

# s1 → school = XYZ (from class)
# s2 → school = XYZ (from class)
# Student → school = XYZ
```

### Instance Methods vs Class Methods vs Static Methods

```python
# Class Method :
class Student:
    school = "ABC"
    
    @classmethod
    def show_school(cls):
        print(cls.school)

# Notice:
# cls

# instead of:
# self

# cls refers to the class itself.

# So:
# Student.show_school()

# roughly becomes:
# Student.show_school(Student)

# Therefore:
# cls → Student class

# and:
# cls.school

# means:
# Student.school
```

### self vs cls — Important

```python
# This is worth locking in before moving forward.
```

```python
# Instance method
class Student:
    def show(self):
        print(self.name)

# Called using an object:
# s1.show()

# Conceptually:
# Student.show(s1)

# So:
# self → s1
```

```python
# Class method
class Student:
    @classmethod
    def show_school(cls):
        print(cls.school)

# Called using the class:
# Student.show_school()

# Conceptually:
# Student.show_school(Student)

# So:
# cls → Student
```

```python
#  Think of it like:
# self → "this particular object"
# cls  → "the entire class"
```

### Static Methods

```python
# A static method is different from both instance and class methods.
```

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
# Notice:

# def add(a, b):

# There is no self and no cls.

# You can call it directly through the class:

print(Math.add(10, 20))
```

```python
# Why use a static method?
# When a function logically belongs to a class but doesn't need information from an object or the class.
```

```python
class Calculator:
    
    @staticmethod
    def multiply(a,b):
        return a*b
    
    @classmethod
    def name(cls):
        return cls.__name__
    
print(Calculator.multiply(5,10))
print(Calculator.name())
```
