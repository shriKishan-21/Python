### Abstraction

```python
# Now abstraction is about hiding unnecessary implementation details and exposing only what the user needs.
```

```python
# 1. Abstract Base Class

# Python provides the abc module for this.

# from abc import ABC, abstractmethod
```

```python
from abc import ABC,abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
```

```python
# Here Animal says:

# Every concrete animal must have a speak() method.

# But Animal doesn't specify exactly how speak() works.
```

```python
# 2. Child classes implement it.
class Dog(Animal):

    def speak(self):
        print("Woof")


class Cat(Animal):

    def speak(self):
        print("Meow")
```

```python
d = Dog()
c = Cat()

d.speak()
c.speak()
```

```python
# The abstract class defines the requirement:

# Animal
#    │
#    └── speak() ← must exist
#        ↑
#    ┌───┴────┐
#    Dog     Cat
#    │        │
#   Woof     Meow
```

```python
# 3. You cannot normally create the abstract class

# Because Animal has an abstract method:

# Animal()

# will raise:

# TypeError

# Python is essentially saying:

# You haven't provided an implementation for speak(), so I can't create a concrete Animal.

# But:

# Dog()

# works because Dog implemented speak().
```

```python
# 🧠 Why is this useful?

# Imagine you're building a payment system:
```

```python
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass
```

```python
# You could then have:
class CreditCard(Payment):
    def pay(self, amount):
        # credit card logic
        pass


class UPI(Payment):
    def pay(self, amount):
        pass
        # UPI logic
```

```python
# The rest of your application only needs to know:

# payment.pay(1000)

# It doesn't need to know the internal implementation.
```

```python
# One subtle but important point

# An abstract method doesn't necessarily have to contain only pass.

# It can have actual code:

class Animal(ABC):

    @abstractmethod
    def speak(self):
        print("Animal sound")

# The important thing is the @abstractmethod decorator:

# A child class must provide its own implementation before it can be instantiated.
```

```python
# Abstraction vs Encapsulation
# These two are often confused.
```

```python
# Encapsulation 🔐

# Focuses on protecting/controlling data and implementation.

# self.__balance

# and:

# @property
```

```python
# Abstraction 🎭

# Focuses on what must be provided, while hiding implementation details.
@abstractmethod
def pay(self):
    pass
```

```python
# Encapsulation
# → "Don't directly mess with my internal data."

# Abstraction
# → "Here's what you need to use; you don't need to know how it works."
```
