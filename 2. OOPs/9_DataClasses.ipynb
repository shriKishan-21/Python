# DataClasses

```python
# Python provides @dataclass to reduce repetitive class code.
```

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student({self.name}, {self.age})"
```

```python
# With a dataclass:
from dataclasses import dataclass

@dataclass
class Student:
    name : str
    age : int
```

```python
# Python automatically gives you useful methods such as:

# __init__
# __repr__
# __eq__
```

```python
s1 = Student("Shri",21)
s2 = Student("Nikhil",22)

print(s1)
print(s1==s2)
```

```python
# 🧠 Why dataclasses?

# Instead of writing:

# __init__
# __repr__
# __eq__

# yourself for a class that mainly stores data, you can simply write:

# @dataclass
# class Student:
#     name: str
#     age: int

# This is extremely common in modern Python.
```

```python
# Why
```

```python
# Without @dataclass

# Suppose you have:

class Student:

    def __init__(self, name, age, college):
        self.name = name
        self.age = age
        self.college = college

    def __repr__(self):
        return f"Student(name={self.name!r}, age={self.age!r}, college={self.college!r})"

    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented

        return (
            self.name == other.name
            and self.age == other.age
            and self.college == other.college
        )

# This is a lot of code just to say:

# "A Student has three pieces of data: name, age, and college."

# Now imagine adding another field:

# address

# You have to remember to update three places:

# __init__    → add address
# __repr__    → add address
# __eq__      → add address

# That's the repetitive work.
```

```python
# With @dataclass

# You simply write:

from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    college: str

# That's it.

# Python generates the common methods for you.

# Conceptually, it gives you something equivalent to:

# def __init__(self, name, age, college):
#     self.name = name
#     self.age = age
#     self.college = college

# and an appropriate:
# __repr__()

# and:
# __eq__()

# So instead of writing roughly 20 lines of boilerplate, you're describing the data directly:

# @dataclass
# class Student:
#     name: str
#     age: int
#     college: str
```

```python
# But there's another BIG reason

# Dataclasses aren't just about saving lines.

# They make your intent obvious.

# Compare:

# class Student:

#     def __init__(self, name, age, college):
#         ...

# with:

# @dataclass
# class Student:
#     name: str
#     age: int
#     college: str

# The second version immediately communicates:

# "This class primarily represents/stores data."

# That's what dataclasses are designed for.
```

```python
# Good use case

# Classes primarily representing data:
# Student
# Employee
# Product
# Point
# Configuration
# User
# Transaction
# Model output


# Less useful when
# Your class is primarily about complex behavior and has little stored data.

# For example:
# class NeuralNetwork:
#     def train(self):
#         ...

#     def predict(self):
#         ...

#     def backpropagate(self):
#         ...

# You probably wouldn't choose a dataclass just because the class happens to have a few attributes.
```

```python
# Normal class:

# "I manually tell Python how to initialize, represent, and compare my data."

# Dataclass:

# "Here are my fields. Python, generate the standard data-handling code for me."
```

### 1. Default values

```python
# You can give fields defaults:

from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int = 18

# Now:

s1 = Student("Shri", 21)
s2 = Student("Rahul")

# Results:

# s1 → Student(name='Shri', age=21)
# s2 → Student(name='Rahul', age=18)

# So age is optional when creating the object.
```

### 2. field() for more control

```python
from dataclasses import dataclass,field

@dataclass
class Student:
    name : str
    marks : list = field(default_factory=list)
    
s1 = Student("Shri")
s2 = Student("Rahul")

s1.marks.append(90)

print(s1.marks)
print(s2.marks)
    
```

```python
# Why default_factory?

# If you wrote:

# marks: list = []

# you could accidentally have multiple objects sharing the same mutable list.

# default_factory=list creates a new list for every object.
```

### 3. Frozen dataclasses 🔒

```python
# You can make a dataclass immutable:

@dataclass(frozen=True)
class Point:
    x: int
    y: int

# Then:
p = Point(10, 20)

p.x = 50

# raises an error.

# You can still create:

# p = Point(50, 20)

# but you can't modify the existing object.

# This is useful when you want objects whose values shouldn't change after creation.
```

### 4. Custom methods still work

```python
# A dataclass is still a normal Python class.

@dataclass
class Rectangle:
    width: int
    height: int

    def area(self):
        return self.width * self.height

# Then:

r = Rectangle(10, 5)

print(r.area())
```

```python
# So dataclass doesn't replace OOP.

# It simply removes boilerplate data-handling code.
```
