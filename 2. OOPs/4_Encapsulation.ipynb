### Encapsulation 🔐

```python
# Encapsulation is about controlling access to an object's data and behavior.

# Python doesn't enforce private variables as strictly as languages like Java/C++, but it provides conventions and name-mangling.
```

```python
class BankAccount:

    def __init__(self, balance):
        self._balance = balance
```

```python
# The single underscore:

# _balance

# means:

# This is intended for internal use.

# It's a convention, not a strict restriction.
```

```python
# You technically can still do:

# account._balance
```

```python
# Double underscore

# Now:

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance
```

```python
# The __balance triggers name mangling.

# Python internally changes the attribute name roughly to:

# __balance
#     ↓
# _BankAccount__balance

# So:
# account.__balance

# will generally fail with:
# AttributeError

# But this is still not true security. Python is saying:
# "This is intended to be private/internal."
```

```python
# Suppose:
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

# Then users interact through methods:

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())

# instead of directly modifying the internal balance.

# This gives us more control over how the data is changed.
```

```python
# __balance is store as _BankAccount__balance internally
# so calling it outside the class will give Attribute error
```

```python
# Properties — The Better Way to Control Attributes 🔥
```

```python
# Instead of:

# account.get_balance()

# we can make the method behave like an attribute:

# account.balance

# using @property.
```

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
        
    @property
    def balance(self):
        return self.__balance
```

```python
account = BankAccount(1000)
print(account.balance)
```

##### The @property decorator makes a method behave like an attribute.

```python
# Without @property

# You might write:

# p.get_name()
# With @property

# You can write:

# p.name

# But internally Python still runs the method:

# p.name
#  ↓
# @property method
#  ↓
# return self.__name
#  ↓
# "Shri"

# So @property gives you attribute-like access with method control.
```

### Property Setter 🔥

```python
# We can also control what happens when someone tries to change the property.
```

```python
class Person:
    def __init__(self,age):
        self.__age = age
      
    @property  
    def age(self):
        return self.__age
    
    @age.setter
    def age(self,value):
        if value<0:
            raise ValueError("Age cannot be negative")
        self.__age = value
```

```python
p = Person(21)
print(p.age)
```

```python
p.age = 25
print(p.age)
```

```python
p.age = -5
```

### Read-only Properties 🔒

```python
# If we define only the getter:
class Person:

    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

# Then:

p = Person(20)

print(p.age)

# works.
```

```python
# But:
p.age = 25
# ❌ causes an AttributeError, because we haven't defined an @age.setter.
```

```python
# So we have a read-only property from the user's perspective.
# p.age       → ✅ read
# p.age = 25  → ❌ write
```
