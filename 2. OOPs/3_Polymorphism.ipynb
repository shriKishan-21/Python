### Polymorphism 🔥

```python
# Different objects can respond to the same method call in their own way.
```

```python
class Dog:
    def speak(self):
        print("Woof")


class Cat:
    def speak(self):
        print("Meow")
```

```python
dog = Dog()
cat = Cat()

dog.speak()
cat.speak()
```

```python
# Both objects have:

# speak()

# but they behave differently.
```

```python
# The really useful part

# We can write:

# def make_sound(animal):
#     animal.speak()

# Then:
  
# make_sound(dog)
# make_sound(cat)

# Output:

# Woof
# Meow

# make_sound() doesn't care whether animal is a Dog or Cat.

# It only cares:

# Does this object have a speak() method?

# This is often called duck typing in Python:
# If it behaves like the thing we need, we can use it
```
