### Dunder / Magic Methods 🪄

```python
# They're called dunder methods because they have double underscores on both sides.
# They allow your class to interact naturally with Python's built-in operations.
```

```python
# 1. __str__

# Consider:

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


s = Student("Shri", 21)

print(s)

# Without __str__, Python doesn't know how you want your object represented to a user, so you'll typically get something like:
# <__main__.Student object at 0x...>
# Not very useful.
```

```python
# We can define:

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} - {self.age}"
    
    
s = Student("Shri", 21)

print(s)
```

```python
# 2. __repr__ — Developer Representation

# __str__ is mainly for a user-friendly representation.
# __repr__ is intended to give a more precise/developer-friendly representation.

# Example:
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    def __str__(self):
        return f"{self.name}, {self.age}"
    
    def __repr__(self):
        return f"Person('{self.name}','{self.age}')"
```

```python
p = Person("Shri",21)
print(p)
```

```python
print(repr(p))
```

```python
#  __str__
# → readable for users

# __repr__
# → useful for developers/debugging
```

```python
# Why does print() prefer __str__?

# When you do:

# print(p)

# Python essentially does:

# print(str(p))

# and str(p) looks for:

# p.__str__()

# But if you don't define __str__, Python can fall back to __repr__.
```

```python
# 3. __len__ 📏
# We can control what happens when Python's len() is used on our object.
```

```python
# Normally:

numbers = [10, 20, 30]
print(len(numbers))

# gives:

# 3

# Python is essentially using:

# numbers.__len__()
```

```python
class length:
    
    def __init__(self):
        self.members = [1,2,3,4]
        
    def __len__(self):
        return len(self.members)
    
    
d = length()
print(len(d))
```

```python
# 4. __eq__ — Comparing Objects

# Here's an important problem with custom objects.
```

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Shri")
s2 = Student("Shri")

print(s1 == s2)
```

```python
#  you'll get:
# False

# Why?
# Because Python doesn't automatically know that two separate Student objects should be considered equal.

# We can define that ourselves with:
# __eq__
```

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        return self.name == other.name
```

```python
s1 = Student("Shri")
s2 = Student("Shri")
print(s1==s2)
```

```python
# 5. __add__ — Making + Work with Your Objects
# You can control what happens when your custom objects use +.
```

```python
class Number:
    def __init__(self,value):
        self.value = value
        
    def __add__(self,other):
        return self.value + other.value
```

```python
d1 = Number(20)
d2 = Number(30)
c = d1+d2
print(c)
```

```python
# Python essentially does:

# c = a.__add__(b)

# So c.value becomes:

# 10 + 20 = 30

# This is called operator overloading.
```

```python
# 7. __getitem__ — Making Objects Work Like Lists 🔥

# Another very useful dunder method is:
# __getitem__

# It controls what happens when you use square brackets:
# obj[index]
```

```python
class Team:
    def __init__(self):
        self.members = [1,2,3,4]
        
    def __getitem__(self,index):
        return self.members[index]
```

```python
team = Team()

print(team[0])
print(team[2])
```

```python
# 8. Making Objects Work with for Loops
# You already learned the iterator protocol:

# __iter__()
# __next__()
```

```python
class Numbers:

    def __init__(self):
        self.data = [10, 20, 30]

    def __iter__(self):
        return iter(self.data)
```

```python
nums = Numbers()

for x in nums:
    print(x)
```

```python
class Numbers:
    
    def __init__(self):
        self.data = [10,20,30]
        self.index = 0
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration
        
        value = self.data[self.index]
        self.index += 1
        return value
```

```python
nums = Numbers()

for num in nums:
    print(num)
```

```python
# for x in nums
#        ↓
# iter(nums)
#        ↓
# nums.__iter__()
#        ↓
# returns nums
#        ↓
# ┌─────────────────────┐
# │   next(nums)        │
# │       ↓             │
# │ __next__()          │
# │       ↓             │
# │ return 10           │
# │                     │
# │   next(nums)        │
# │       ↓             │
# │ return 20           │
# │                     │
# │   next(nums)        │
# │       ↓             │
# │ return 30           │
# │                     │
# │   next(nums)        │
# │       ↓             │
# │ StopIteration       │
# └─────────────────────┘
#        ↓
#      STOP
```

```python
# 9. __call__

# This one is especially interesting because it makes an object behave like a function.
```

```python
class Multiplier:

    def __init__(self, n):
        self.n = n

    def __call__(self, x):
        return x * self.n


double = Multiplier(2)

print(double(10))
print(double(5))
```

```python
# Why is this useful?

# A normal function:

# def double(x):
#     return x * 2

# stores no persistent state.

# But our object:

# double = Multiplier(2)

# stores:

# double
#  └── n → 2

# and can still be called like a function:

# double(10)

# This pattern is very common in Python.
```

```python
# 10. __bool__ 🔥

# We can control what happens when Python asks:

# if obj:

# Normally, Python determines whether an object is truthy or falsy.
```

```python
class Account:

    def __init__(self, balance):
        self.balance = balance

    def __bool__(self):
        return self.balance > 0
```

```python
a = Account(500)
b = Account(0)

print(bool(a))
print(bool(b))
```

```python
class Box:

    def __init__(self, items):
        self.items = items

    def __bool__(self):
        return len(self.items) > 0


a = Box([10, 20])
b = Box([])

print(bool(a))
print(bool(b))

if a:
    print("A has items")

if b:
    print("B has items")
```

```python
# 11. __contains__ — The in Operator

# We can control:

# x in obj

# using:

# __contains__
```

```python
class Team:
    def __init__(self):
        self.members = [10,20,30]
        
    def __contains__(self, names):
        return names in self.members
```

```python
a = Team()
print(10 in a)
print(20 in a)
print(40 in a)
```

```python
# 12. __setitem__ — Making obj[index] = value Work
# We already learned:
# obj[index]

# uses:
# __getitem__

# Similarly:
# obj[index] = value

# uses:
# __setitem__
```

```python
class Team:
    def __init__(self):
        self.members = [10,20,30]
        
    def __getitem__(self, key):
        return self.members[key]
    
        
    def __setitem__(self, key, value):
        self.members[key] = value
        
d = Team()
d[1] = 30
d[2] = 40
print(d[1])
print(d[2])
```

```python
# 13. __delitem__
class Box:

    def __init__(self):
        self.data = [10, 20, 30]

    def __getitem__(self, index):
        return self.data[index]

    def __setitem__(self, index, value):
        self.data[index] = value

    def __delitem__(self, index):
        del self.data[index]
```

```python
box = Box()

del box[1]

print(box.data)
```

```python
# after the first deletion, the indexes change.
```

```python
# 14. __enter__ and __exit__ — Context Managers

# This connects directly to the with statement you've already encountered.
```

```python
# Normally:

# with open("file.txt") as f:
#    data = f.read()

# Python needs to:

# enter → do the work → exit/cleanup

# We can create our own with-compatible class using:
# __enter__()
# __exit__()
```

```python
class Demo:

    def __enter__(self):
        print("Enter")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exit")


with Demo():
    print("Inside")
```

```python
# And importantly, __exit__() runs even if an exception occurs inside the with block.
```

```python
# __exit__ and exceptions

# Remember, __exit__ receives three arguments:

# def __exit__(self, exc_type, exc_value, traceback):

# If there is no exception:
# exc_type  → None
# exc_value → None
# traceback → None

# If an exception occurs, they contain information about it.
```

```python
class Demo:

    def __enter__(self):
        print("Enter")

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exit")
        print(exc_type)


with Demo():
    print("Inside")
    raise ValueError("Something went wrong")
```

```python
# __exit__ still runs even though an exception occurred.

# That's why context managers are useful for cleanup.
```

### __new__ vs __init__

```python
# You've been writing:

# obj = MyClass(...)

# and we've treated that as:

# create object
#    ↓
# __init__()

# But there's actually another step before __init__.

# The real flow is roughly:

# MyClass(...)
#    ↓
# __new__()
#    ↓
# object is CREATED
#    ↓
# __init__()
#    ↓
# object is INITIALIZED
```

```python
# __new__
# Responsible for creating/allocating the object.

# __init__
# Responsible for initializing the already-created object.
```

```python
class Person:

    def __new__(cls):
        print("NEW")
        return super().__new__(cls)

    def __init__(self):
        print("INIT")


p = Person()
```

```python
# __new__ with immutable objects

# This is where __new__ becomes genuinely useful.

# Consider integers:

# x = 10

# An integer is immutable — you can't modify the existing 10 object.

# That's why when working with immutable types like:

# int
# str
# tuple

# __new__ can be used to control how the object is created.
```

```python
# 18. __del__ — Object Destruction

# Python has a dunder method called:
# __del__

# It can run when an object is being destroyed.

# Example:

# class Demo:

def __del__(self):
    print("Object destroyed")


d = Demo()

del d
```

```python
# Output:
# Object destroyed

# Here:
# del d

# removes the name d's reference to the object. If that was the object's last reference, Python can destroy the object, and __del__ may run.

# Important ⚠️
# Don't think of __del__ as a guaranteed, immediate cleanup mechanism.

# Python's garbage collection and object lifetime mean you shouldn't rely on the exact timing of __del__.

# For resources like:
# files
# database connections
# network connections

# use a context manager (with) instead.
```
