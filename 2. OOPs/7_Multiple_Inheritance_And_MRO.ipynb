```python
# Python allows a class to inherit from multiple classes:

class A:
    def show(self):
        print("A")


class B:
    def show(self):
        print("B")


class C(A, B):
    pass

# Now:

c = C()
c.show()
```

```python
# What does Python do?

# It doesn't randomly choose between A and B.

# It follows the MRO — Method Resolution Order.
```

```python
# the MRO is:

# C → A → B → object

# So Python searches:

# C
# ↓
# A  ← show found ✅
# ↓
# STOP

# Therefore:

# A
```

```python
# You can see it yourself
# print(C.mro())

# You'll get conceptually:

# [C, A, B, object]

# object is Python's ultimate base class for normal classes.
```

```python
print(C.mro())
```

```python
# Now change the order
class C(B, A):
    pass

# Now the MRO becomes:

# C → B → A → object

# Therefore:

c.show()

# finds show() in B first.

# Output:
# B
```

```python
# 🧠 Key rule

# The order inside:

# class C(A, B):

# matters.

# C(A, B)
#  ↓
# C → A → B → object

# C(B, A)
#  ↓
# C → B → A → object
```

```python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")
        super().show()


class C(A):
    def show(self):
        print("C")
        super().show()


class D(B, C):
    pass

d = D()
d.show()
```

```python
# The MRO of D is:

# D → B → C → A → object

# That's the crucial part.

# Now:

# d.show()

# Step 1
# Python searches:
# D → B

# Finds B.show():
# B
```

```python
# Step 2

# Inside B.show():
# super().show()

# ⚠️ super() does not simply mean "call B's parent A".
# It means:
# Continue searching after B in D's MRO.

# MRO:
# D → B → C → A → object
#       ↑
#       current

# After B comes C.

# So:
# super().show()

# calls:

# C.show()

# Therefore:
# C
```

```python

# Step 3

# Inside C.show():
# super().show()

# Continue after C:
# D → B → C → A → object
#               ↑

# So it calls:
# A.show()

# →
# A

# Final:
# B
# C
# A
```

```python
# 🔥 The key idea

# Never think:

# super() = my direct parent

# Think:

# super() = continue from the next class in the MRO

# This distinction is extremely important in Python multiple inheritance.
```
