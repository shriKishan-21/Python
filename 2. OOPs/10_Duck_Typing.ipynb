### Duck Typing 🦆

##### The philosophy is:
##### "If it behaves like a duck, treat it like a duck."

```python
class Dog:
    def speak(self):
        print("Woof")


class Cat:
    def speak(self):
        print("Meow")
```

```python
def make_sound(animal):
    animal.speak()
    
make_sound(Dog())
make_sound(Cat())
```

```python
# Notice something important:

# We never wrote:

# def make_sound(animal: Animal):

# and neither Dog nor Cat needs to inherit from a common class.

# Python simply cares:

# Does the object have a speak() method?

# This is duck typing.

# Dog → speak() ✅
# Cat → speak() ✅

# make_sound()
#      ↓
# animal.speak()

# The object's behavior matters more than its class hierarchy.
```

### Protocols — Formalizing Duck Typing

```python
# Duck typing works great, but sometimes we want to describe the expected interface more formally.

# Python provides Protocol for this.

from typing import Protocol

class Speaker(Protocol):

    def speak(self):
        ...
```

```python
class Dog:

    def speak(self):
        print("Woof")


class Robot:

    def speak(self):
        print("Beep")
```

```python
# Neither class needs to inherit from Speaker.

# We can write:

def make_sound(obj: Speaker):
    obj.speak()

# And both work:

make_sound(Dog())
make_sound(Robot())

# This is called structural typing.
```

```python
# The difference

# Traditional inheritance asks:
# "Does this class inherit from the required class?"

# Duck typing / Protocol asks:
# "Does this object have the required behavior?"
```

```python
# Inheritance:
# Dog → Animal

# Structural typing:
# Dog → has speak() → usable as Speaker
# Robot → has speak() → usable as Speaker

# This is especially useful in larger Python applications because you can design code around interfaces/behavior rather than rigid inheritance hierarchies.
```
