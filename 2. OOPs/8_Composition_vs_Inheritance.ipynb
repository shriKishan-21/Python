# Composition vs Inheritance 🔥

```python
# Inheritance = "is-a"
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    pass

# A Dog is an Animal.

# Dog → is-a → Animal

# So inheritance makes sense.
```

```python
# Composition = "has-a"

# Instead of inheriting, one object contains another object.

# Example:

class Engine:

    def start(self):
        print("Engine started")


class Car:

    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()
        print("Car started")

# A Car has an Engine.

# Car
# └── engine → Engine object

# Now:

car = Car()
car.start()

# Output:

# Engine started
# Car started
```

```python
# The important relationship is:

# Inheritance:
# Dog IS-A Animal

# Composition:
# Car HAS-A Engine
```

```python
# Why composition is powerful

# Suppose you have:

class Engine:
    def start(self):
        print("Engine")


class ElectricMotor:
    def start(self):
        print("Motor")

# You can make different cars using different components:

class Car:

    def __init__(self, power_source):
        self.power_source = power_source

    def start(self):
        self.power_source.start()

# Then:
```

```python
car1 = Car(Engine())
car2 = Car(ElectricMotor())

car1.start()
car2.start()
```

```python
# 🧠 Rule of thumb

# Ask:

# Is the child actually a type of the parent?

# If yes:

# "is-a" → inheritance

# If instead:

# Does this object use/contain another object?

# Then:

# "has-a" → composition
```

```python
# Dependency Injection
```

```python
class Battery:
    def charge(self):
        print("Charging")


class Phone:
    def __init__(self, battery):
        self.battery = battery

    def charge(self):
        self.battery.charge()


battery = Battery()
phone = Phone(battery)

phone.charge()
```

```python
# The important part is:

# phone = Phone(battery)

# Phone receives its dependency instead of creating it itself.

# This makes the class more flexible and easier to test.
```
