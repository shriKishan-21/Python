```python
# 1. Traversals

# Create this tree:

#         10
#        /  \
#       5    15
#      / \   / \
#     2   7 12 20

# Implement:

# preorder(root)
# inorder(root)
# postorder(root)
# level_order(root)
```

```python
from collections import deque
class BinarySearchTree:
    def __init__(self,value):
        self.value = value
        self.left = None
        self.right = None
    
# insertion
def insertion(root,val):
    if root is None:
        return BinarySearchTree(val)
    
    if root.value > val:
        root.left = insertion(root.left,val)
    else:
        root.right = insertion(root.right,val)
    
    return root

# inorder
def inorder(root):
    if root is None:
        return 
    
    inorder(root.left)
    print(root.value,end=" ")
    inorder(root.right)
    
# preorder
def preorder(root):
    if root is None:
        return
    
    print(root.value,end=" ")
    preorder(root.left)
    preorder(root.right)
    
# postorder
def postorder(root):
    if root is None:
        return
    
    postorder(root.left)
    postorder(root.right)
    print(root.value,end=" ")
    
# level order traversal
def level_order_traversal(root):
    if root is None:
        return
    
    queue = deque()
    queue.append(root)
    
    while queue:
        node = queue.popleft()
        
        print(node.value,end=" ")
        
        if node.left:
            queue.append(node.left)
            
        if node.right:
            queue.append(node.right)
    
    
```

```python
root = BinarySearchTree(10)
insertion(root,5)
insertion(root,15)
insertion(root,2)
insertion(root,7)
insertion(root,12)
insertion(root,20)
inorder(root)
print()
preorder(root)
print()
postorder(root)
print()
level_order_traversal(root)
```

```python
# 2. Height

# Implement:

# height(root)

# For the tree above:
# 3
```

```python
def height(root):
    if root is None:
        return 0
    
    left = height(root.left)
    right = height(root.right)
    
    return max(left,right)+1
```

```python
# 3. Count Nodes

# Implement:
# count_nodes(root)

# For the tree above:
# 7

# Try recursion.
```

```python
def count_nodes(root):
    if root is None:
        return 0
    
    left = count_nodes(root.left)
    right = count_nodes(root.right)
    
    return left+right+1
```

```python
# 4. Search in BST

# Implement:

# search(root, target)

# Test:

# 12 → True
# 19 → False
```

```python
def search(root,target):
    if root is None:
        return False
    
    if root.value == target:
        return True
    
    if root.value > target:
        return search(root.left,target)
    return search(root.right,target)
```

```python
# 5. Validate BST

# Given a binary tree, determine whether it is actually a valid BST.

# Don't only compare:
# root.left < root
# root.right > root

# because that's insufficient for deeper nodes.

# Example:

#         10
#        /  \
#       5    15
#           /
#          6

# This is not a valid BST because 6 is in the right subtree of 10 but is smaller than 10.

# This is an important interview problem.
```

```python
def inorder(root,prev):
    if root is None:
        return False
    # L
    inorder(root.left,prev)
    # N
    if prev>=root.value:
        return True
    prev = root.value
    # R
    inorder(root.right,prev)

def validate_bst(root):
    return inorder(root,root.value) 
```

```python
validate_bst(root)
```

```python
# 6. Kth Smallest in BST

# Given:

#         8
#        / \
#       4   12
#      / \  / \
#     2  6 10 14

# Find the:

# 3rd smallest

# Expected:

# 6

# Hint:

# What order does inorder traversal produce?
```

```python
import heapq

def k_smallest(order,k):
    heap = []
    
    for val in order:
        heapq.heappush(heap,-val)
        if len(heap)>k:
            heapq.heappop(heap)
    
    return [-x for x in heap]
    
def inorder(root):
    if root is None:
        return []
    
    order = []
    
    order += inorder(root.left)
    order.append(root.value)
    order += inorder(root.right)
    
    return order

# TC : O(nlogk)
# Sc : O(n)

# approach 2
def k_smallest_2(root,k):
    result = []
    
    def reverse_inorder(node):
        if node is None or len(result)==k:
            return 
        
        # smallest element first
        reverse_inorder(node.left)
        
        # save it
        if len(result)<k:
            result.append(node.value)
            
        reverse_inorder(node.right)
    
    reverse_inorder(root)
    return result
            
```

```python
order = inorder(root)
print(order)
print(k_smallest(order,2))
k_smallest_2(root,2)
```

```python
# 7. Level Order by Levels

# Instead of:

# 10 5 15 2 7 12 20

# return:

# [
#     [10],
#     [5, 15],
#     [2, 7, 12, 20]
# ]

# Use a queue.
```

```python
def level_order_traversal_level(root):
    if root is None:
        return []
    
    result = []
    queue = deque([root])
    
    while queue:
        size = len(queue)
        current = []
        
        for _ in range(size):
            node = queue.popleft()
            
            current.append(node.value)
            
            if node.left:
                queue.append(node.left)
                
            if node.right:
                queue.append(node.right)
                
        result.append(current)
        
        
    return result
```

```python
level_order_traversal_level(root)
```

```python
# 8. K Largest Elements

# Given:

# nums = [3, 1, 5, 2, 4, 8, 7]
# k = 3

# Return the 3 largest elements.

# Try solving it using a heap of size k, rather than sorting the entire array.
```

```python
import heapq

def k_largest(order,k):
    heap = []
    
    for val in order:
        heapq.heappush(heap,val)
        if len(heap)>k:
            heapq.heappop(heap)
    
    return heap
    
def inorder(root):
    if root is None:
        return []
    
    order = []
    
    order += inorder(root.left)
    order.append(root.value)
    order += inorder(root.right)
    
    return order

# TC : O(nlogk)
# Sc : O(n)

# approach 2
def k_largest_2(root,k):
    result = []
    
    def reverse_inorder(node):
        if node is None or len(result)==k:
            return 
        
        # Largest element first
        reverse_inorder(node.right)
        
        # save it
        if len(result)<k:
            result.append(node.value)
            
        reverse_inorder(node.left)
    
    reverse_inorder(root)
    return result
            
```

```python
order = inorder(root)
print(order)
k_largest(order,2)
```

```python
k_largest_2(root,3)
```

```python
# 9. Implement a Trie

# Create:

# class TrieNode:
#     ...

# class Trie:
#     ...

#     def insert(self, word):
#         ...

#     def search(self, word):
#         ...

#     def starts_with(self, prefix):
#         ...

# Test:

# trie.insert("apple")
# trie.insert("app")
# trie.insert("application")

# Then:

# trie.search("app")          # True
# trie.search("appl")        # False
# trie.starts_with("appl")   # True
# trie.starts_with("xyz")    # False
```

```python
class Trieroot:
    def __init__(self):
        self.children = {}
        self.is_end = False
        
        
class Trie:
    def __init__(self):
        self.root = Trieroot()
        
    def insert(self,word):
        current = self.root
        
        for ch in word:
            if ch not in current.children:
                current.children[ch] = Trieroot()
                
            current = current.children[ch]
        
        current.is_end = True
                
    def search(self,word):
        current = self.root
        
        for ch in word:
            if ch not in current.children:
                return False
            
            current = current.children[ch]
        
        return current.is_end
        
    def starts_with(self,word):
        current = self.root
        
        for ch in word:
            if ch not in current.children:
                return False
            
            current = current.children[ch]
            
        return True
```

```python
trie = Trie()
trie.insert("apple")
trie.insert("app")
trie.insert("application")
```

```python
print(trie.search("app"))   
print(trie.search("appl"))   
print(trie.starts_with("appl"))   
print(trie.starts_with("xyz"))   
```
