```python
# 3. Linked List

# Create:
# class Node
# class LinkedList

# Implement:
# insert_at_beginning()
# insert_at_end()
# display()
# search()
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None
        
    def insert_at_beginning(self,value):
        newNode = Node(value)
        
        newNode.next = self.head
        self.head = newNode
        
    def insert_at_end(self,value):
        newNode = Node(value)
        
        if self.head is None:
            self.head = newNode
            return
        
        
        current = self.head
        while current.next:
            current = current.next
        
        current.next=newNode
        
    def display(self):
        current = self.head
        
        while current:
            print(current.data,end="->")
            current = current.next
        print("Null")
    
    def search(self,val):
        current = self.head
        while current.next:
            if current.data == val:
                print("Found")
                return
            current=current.next
        print("Not Found")     
        
```

```python
ll = LinkedList()
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.insert_at_end(40)
ll.display()
ll.insert_at_beginning(10)
ll.display()
ll.search(30)
ll.search(50)
```

```python
# 4. Reverse a Linked List

# Input:
# 1 → 2 → 3 → 4 → None

# Output:
# 4 → 3 → 2 → 1 → None

# Try the iterative version first.
# Then try recursion.
```

```python
def reverse_Linkedlist(head):
        if head is None or head.next is None:
            return head
        
        curr = head
        prev = None
        while curr:
            forw = curr.next
            curr.next = prev
            prev=curr
            curr=forw
        return prev 
```

```python
def solve(self,curr,prev):
    if curr is None:
        return prev

    forw = curr.next
    curr.next=prev

    return self.solve(forw,curr)

def reverseList(self, head):
    return self.solve(head,None)
```

```python
# 5. Middle of Linked List

# Input:
# 1 → 2 → 3 → 4 → 5

# Return:
# 3

# Hint:
# Slow pointer + fast pointer
```

```python
def middle_of_linkedlist(self,head):
    # find length
    len = 0
    curr = head
    while curr:
        len+=1
        curr = curr.next
    
    mid =len//2
    
    curr = head
    for _ in range(mid):
        curr = curr.next
    
    return curr

def middle_ll(self,head):
    slow = head
    fast = head
    while fast and fast.next:
        fast = fast.next.next
        slow = slow.next
    return slow
```

```python
def binary_Search(nums,target):
    s = 0
    e = len(nums)
    while s<=e:
        mid = s+(e-s)//2
        if nums[mid]==target:
            return mid
        elif nums[mid]>target:
            e=mid-1
        else:
            s=mid+1
    # Does not found
    return -1
```

```python
# 7. Valid Parentheses

# Given:
# "()[]{}"

# return:
# True

# Given:
# "([)]"

# return:
# False
```

```python
def valid_parenthese(s):
    stack = []
    if len(s)==0 or len(s)==1:
        return False
    for char in s:
        if char=='(' or char=='[' or char=='{':
            stack.append(char)
        else:
            if len(stack)==0:
                return False
            ch = stack[-1]
            stack.pop()
            if (char==')' and ch!='(') or (char=='}' and ch!='{') or (char==']' and ch!='['):
                return False

    if len(stack)!=0:
        return False
    return True
```

```python
str = "([){})"
print(valid_parenthese(str))
```

```python
# 8. First Occurrence

# Given:
# nums = [1, 2, 2, 2, 4, 5]
# target = 2

# return:
# 1

# Use binary search, not linear search.
```

```python
def first_occurrence(nums,target):
    s = 0
    e = len(nums)
    result = 0
    while s<=e:
        mid = s+(e-s)//2
        
        if nums[mid]==target:
            result = mid
            e = mid-1
        elif nums[mid]>target:
            e = mid-1
        else:
            s=mid+1
    return result
```

```python
nums = [1,2,2,2,4,5]
target = 2
print(first_occurrence(nums,target))
```

```python
# 9. Sort Without sort()

# Implement insertion sort for:
# nums = [5, 3, 8, 1, 2]

# Expected:
# [1, 2, 3, 5, 8]
```

```python
def insertion_sort(nums):
    for i in range(1,len(nums)):
        key = nums[i]
        j = i-1
        while j>=0 and nums[j]>key:
            nums[j+1] = nums[j]
            j-=1
        nums[j+1] = key
        
    return nums
```

```python
nums = [5, 3, 8, 1, 2]
insertion_sort(nums)
```

### __🔥 Phase 2 Challenge__

```python
# This one combines multiple concepts.

# 10. Min Stack

# Design a stack supporting:
# push(x)
# pop()
# top()
# get_min()

# Example:
# push(5)
# push(3)
# push(7)
# push(2)

# Then:
# get_min() → 2
# pop()
# get_min() → 3

# The challenge is to make get_min() O(1).
# Think about what extra information you could maintain.
```

```python
class Stack:
    def __init__(self):
        self.st = []
        self.min_st = []
                
    def push(self,value):
        self.st.append(value)   
        
        if not self.min_st or value<=self.min_st[-1]:
            self.min_st.append(value) 
        
    def pop(self):
        value = self.st.pop()
        
        if value == self.min_st[-1]:
            self.min_st.pop()
            
        return value
        
    def top(self):
        return self.st[-1]
    
    def get_min(self):
        return self.min_st[-1]
```

```python
s = Stack()
s.push(5)
s.push(3)
s.push(7)
s.push(2)
print(s.get_min())
print(f"Pop : {s.pop()}")
print(s.get_min())
print(f"Pop : {s.pop()}")
print(s.get_min())
```

Complexity  
push    → O(1)  
pop     → O(1)  
top     → O(1)  
get_min → O(1)  
space   → O(n)
