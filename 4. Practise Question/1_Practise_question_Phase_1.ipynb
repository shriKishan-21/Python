```python
# 2. Reverse Array
nums = [1,2,3,4,5,6]

left = 0
right = len(nums)-1
while left<right:
    nums[left],nums[right] = nums[right],nums[left]
    left+=1
    right-=1 
    
print(nums)
```

```python
# 3. Count Vowels
s = "programming"

count = 0
vowel = ['a','e','i','o','u']
for x in s:
    if x in vowel:
        count+=1
        
print(count)
```

```python
# 5. Maximum Sum of K Elements
def max_sum(nums,k):
    window_sum = sum(nums[:k])
    max_window = window_sum
    
    for right in range(k,len(nums)):
        window_sum += nums[right]
        window_sum -= nums[right-k]
        
        max_window = max(max_window,window_sum)
    return max_window
```

```python
nums = [1,2,3,4,5,6]
max_sum(nums,2)
```

```python
# 8. Two sum
def two_sum(nums,target):
    seen = {}
    
    for i,x in enumerate(nums):
        need = target - x
        
        if need in seen:
            return [seen[need],i]
        
        seen[x]=i
        
nums = [2, 7, 11, 15]
target = 9
two_sum(nums,target)
```
