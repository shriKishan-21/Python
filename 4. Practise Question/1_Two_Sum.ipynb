```python

def two_sum(nums,target):
    seen = {}
    
    for i,x in enumerate(nums):
        need = target - x
        
        if need in seen:
            return [seen[need],i]
        
        seen[x]=i
        
nums = [2, 7, 11, 15]
target = 9
two_sum(nums,target)
```

```python
#  Now:

# Time:  O(n)
# Space: O(n)

# This is one of the most important transformations in DSA:
# Trade memory for speed.
```
