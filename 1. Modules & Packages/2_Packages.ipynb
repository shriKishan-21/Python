```python
# Packages 📦
# A package is basically a directory used to organize multiple modules.
```

```python
# For example:

# myproject/
# │
# ├── main.py
# │
# └── utils/
#     ├── __init__.py
#     ├── math.py
#     └── string.py
```

```python
# Here:

# utils
#  ├── math.py
#  └── string.py

# is a package.
```

```python
# You can import:
# from utils.math import add
```

```python
# One important package detail: __init__.py

# You'll often see:
# tools/
# ├── __init__.py
# ├── math.py
# └── text.py

# Historically, __init__.py was required for Python to recognize a directory as a package.
# Modern Python also supports namespace packages without it, but you'll still see __init__.py very commonly in real projects.
# It can also contain package initialization code or expose selected functions.
# For now, just remember:

# __init__.py
#      ↓
# marks/initializes a package
```
