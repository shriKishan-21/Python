```python
# 1. What is a module?

# A module is simply a Python file (.py) containing code.
```

```python
import math_utils

print(math_utils.sum(10,20))
print(math_utils.subtract(50,20))
print(math_utils.mul(10,2))
```

```python
# 2. from ... import

# Instead of:
import math_utils

math_utils.sum(10, 20)

# you can write:
from math_utils import sum

print(sum(10, 20))
```

```python
# 3. as

# You can give an imported module another name:

import math_utils as m

print(m.sum(10, 20))
```

```python
# __name__ and __main__ 🔥
```

```python
# Consider a file called calculator.py:

def add(a, b):
    return a + b

print("Calculator loaded")
```

```python
# If another file does:
# import calculator

# then the print("Calculator loaded") also runs.

# That's because importing a module executes its top-level code.
```

```python
# To prevent certain code from running when imported, we use:
# if __name__ == "__main__":
```

```python
def add(a, b):
    return a + b


if __name__ == "__main__":
    print(add(10, 20))
```

```python
# Situation 1: Run calculator.py directly

# Python sets:
# __name__ == "__main__"

# So:
# 30
# is printed.
```

```python
# Situation 2: Import calculator.py
# import calculator

# Now:
# __name__ == "calculator"

# So:
# if __name__ == "__main__":
# is False.

# The test code doesn't run.
```

```python
# 🧠 Simple rule

# Run file directly
# → __name__ = "__main__"

# Import file
# → __name__ = module name

# This lets us separate:
# Reusable code
# +
# Code that should run only when the file itself is executed
```
