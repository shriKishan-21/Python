__Unlike an array/list, a linked list stores data in nodes connected by references.__

A node looks like:  
[data | next]

Example:  
10 → 20 → 30 → None

Each node knows where the next node is.

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

```python
a = Node(10)
b = Node(20)
c = Node(30)

a.next = b
b.next = c
```

```python
current = a
while current:
    print(current.data)
    current = current.next
```

```python
# Traversal: O(n)
```

__Insertion in Linked list__

```python
class LinkedListInsertion:
    def __init__(self):
        self.head = None
    
    
    # insert at end
    def insert_end(self,data):
        newNode = Node(data)
        
        if self.head is None:
            self.head = newNode
            return
        
        current = self.head
        
        while current.next:
            current = current.next
            
        current.next = newNode
    
    # remove from end
    def remove_end(self):
                
        if self.head is None:
            return
        
        if self.head.next is None:
            self.head = None
            return None
        
        current = self.head
        while current.next.next:
            current = current.next
        
        current.next = None
        
        
    # insert at beginning
    def insert_start(self,data):
        newNode = Node(data)
        
        newNode.next = self.head
        self.head = newNode
        
    # Remove from beginning
    def remove_start(self):
        
        if self.head is None:
            return None
                
        self.head = self.head.next
        
        
    # insert at given position
    def insert_between(self,data,position):
        newNode = Node(data)
        
        if position==1:
            self.insert_start(data)
            return
        
        if self.head is None:
            return
            
        current = self.head
        pos = 1
        while pos!=position-1:
            if current.next is None:
                return
            current=current.next
            pos+=1
            
        newNode.next=current.next
        current.next = newNode
        
    # remove from given postion
    def remove_between(self,position):        
        if self.head is None:
            return
        
        if position==1:
            self.remove_start()
            return
            
        current = self.head
        pos = 1
        
        while pos < position-1:
            if current.next is None:
                return
            current=current.next
            pos+=1
        
        if current.next is Node:
            return
        
        current.next = current.next.next
                
        
    def display(self):
        current = self.head
        
        while current:
            print(current.data,end="->")
            current = current.next
        print("NULL")
    
```

```python
ll = LinkedListInsertion()

ll.insert_end(10)
ll.insert_end(20)
ll.insert_end(40)

ll.insert_start(00)
ll.insert_between(30,4)
ll.display()

ll.remove_start()
ll.display()

ll.remove_end()
ll.display()

ll.remove_between(2)
ll.display()
```

```python
# Complexity of your methods
# insert_start   → O(1)
# insert_end     → O(n)
# insert_between → O(n) worst case
# remove_start   → O(1) 
# remove_end     → O(n) 
# remove_between → O(n) worst case
# display        → O(n)
```

__Why linked lists?__

Insertion/deletion at a known node can be efficient because we change references rather than shift many elements.

But accessing:  
index 500

requires walking from the beginning.

So:
Array index access → O(1)  
Linked-list index access → O(n)
