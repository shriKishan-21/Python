```python
# Python strings are sequences:

s = "python"
```

```python
# Indexing:

print(s[0])     # p
print(s[-1])    # n
```

```python
# Strings are immutable.
```

```python
# This doesn't work:

s[0] = "P"
```

```python
# Instead:
print(s)
s = "P" + s[1:]
print(s)
```

```python
# Useful operations
print(s.lower())
print(s.upper())
print(s.strip())
print(s.split())
print(s.replace("n", "b"))
```

```python
# And:

"py" in s

# is useful for membership checking.
```

```python
chars = ['a','b','c','d']
```

```python
# Building strings

# Avoid repeatedly doing:

result = ""

for ch in chars:
    result += ch

print(result)
# for large amounts of data.
```

```python
# Prefer:

result = "".join(chars)
print(result)

# This is generally more efficient.
```
