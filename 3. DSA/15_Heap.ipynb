__Heap__

A heap is a complete binary tree used primarily for efficiently retrieving the minimum or maximum element.

<div class = "alert alert-success">

Two types:

Min Heap → smallest at root    
Max Heap → largest at root  

<div class = "alert alert-warning">

Python's built-in __heapq__ is a min heap.

```python
import heapq

heap = []

heapq.heappush(heap,5)
heapq.heappush(heap,2)
heapq.heappush(heap,8)
heapq.heappush(heap,1)

print(heapq.heappop(heap))
```

<div class = "alert alert-info">

Why Heap?

Suppose you repeatedly need:

Give me the smallest element.

Sorting every time would be wasteful.

A heap lets you maintain that structure efficiently.

<div class = "alert alert-success">

Typical operations:

| Operation  | Complexity |
| ---------- | ---------: |
| Peek min   |       O(1) |
| Push       |   O(log n) |
| Pop min    |   O(log n) |
| Build heap |       O(n) |


```python
# Heap Example — Top K
nums = [10, 4, 7, 1, 9, 3]

print(heapq.nlargest(3,nums))
```

```python
# But in DSA, you'll often implement the idea with a small heap of size k:
def top_k_largest(nums,k):
    heap = []
    
    for x in nums:
        heapq.heappush(heap,x)
        if len(heap)>k:
            heapq.heappop(heap)
            
    return heap
```

```python
nums = [10, 4, 7, 1, 9, 3]
top_k_largest(nums,2)
```
