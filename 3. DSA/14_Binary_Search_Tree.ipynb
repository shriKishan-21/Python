A BST has an ordering rule:

left subtree < root < right subtree

```python
class BinarySearchTree:
    def __init__(self,value):
        self.value = value
        self.left = None
        self.right = None
```

```python
# BST Search
def search(root,target):
    if root is None:
        return False
    
    if root.data == target:
        return True
    
    if target<root.data:
        search(root.left,target)
        
    search(root.right,target)
    
    
# Average for a reasonably balanced BST: O(log n)
# Worst case for a completely skewed tree: O(n)
```

```python
# BST Insertion
class BinarySearchTree:
    def __init__(self,value):
        self.value = value
        self.left = None
        self.right = None
        
def insert(root,value):
    if root is None:
        return BinarySearchTree(value)
    
    if value < root.value:
        root.left = insert(root.left,value)
    else:
        root.right = insert(root.right,value)
        
    return root
```

```python
root = None

for value in [8, 4, 12, 2, 6, 10, 14]:
    root = insert(root,value)
```

__BST Deletion__

```python
# Deletion has three cases.

def delete_leaf(root,val):
    if root is None:
        return
    if root.value == val:
        # Case 1: Leaf
        if root.left is None and root.right is None:
            del root
            return None
        # Case 2: One child
        if root.left is not None and root.right is None:
            temp = root.left
            del root
            return temp
        
        if root.left is None and root.right is not None:
            temp = root.right
            del root
            return temp
            
        # Case 3: Two children
        if root.right is not None and root.right is not None:
            pass
        
        
    elif root.value > val:
        root.left = delete_leaf(root.left)
        
    else:
        root.right = delete_leaf(root.right)
    
    return root
```
