```python
# Trie
# A Trie is a tree specifically designed for strings/prefixes.
```

Suppose we store:

cat
car
care

A conceptual trie:

        root
         |
         c
         |
         a
        / \
       t   r
           |
           e

The prefix:

ca

is shared.

That's the main advantage.

```python
# Trie Node
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False
    
class Trie:
    def __init__(self):
        self.root = TrieNode()
```

```python
# Trie Insert
def insert(self,word):
    current = self.root
    
    for ch in word:
        if ch not in current.children:
            current.children[ch] = TrieNode()
        current = current.children[ch] 
    current.is_end = True
```
