```python
# Trie
# A Trie is a tree specifically designed for strings/prefixes.
```

Suppose we store:

cat
car
care

A conceptual trie:

        root
         |
         c
         |
         a
        / \
       t   r
           |
           e

The prefix:

ca

is shared.

That's the main advantage.

```python
# Trie Node
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False
    
class Trie:
    def __init__(self):
        self.root = TrieNode()
```

```python
# Trie Insert
def insert(self,word):
    current = self.root
    
    for ch in word:
        if ch not in current.children:
            current.children[ch] = TrieNode()
        current = current.children[ch] 
    current.is_end = True 
```

```python
# Trie Search
def search(self, word):
    current = self.root

    for ch in word:
        if ch not in current.children:
            return False

        current = current.children[ch]

    return current.is_end
```

```python
# Prefix Search
def starts_with(self, prefix):
    current = self.root

    for ch in prefix:
        if ch not in current.children:
            return False

        current = current.children[ch]

    return True
```

```python
#  Trie Complexity

# For a word of length L:

# Insert → O(L)
# Search → O(L)
# Prefix → O(L)

# This is excellent when operations depend on prefixes.
```
