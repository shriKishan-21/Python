```python
# Prefix sums allow us to answer range-sum queries quickly.
```

```python
# Given:

# nums = [2, 4, 1, 5, 3]

# Build:

# prefix = [0, 2, 6, 7, 12, 15]

# Meaning:

# prefix[i] = sum of elements before i

# So:

# sum(1..3)

# which means:

# 4 + 1 + 5 = 10

# can be calculated:

# prefix[4] - prefix[1]
# = 12 - 2
# = 10
```

```python
def build_prefix(nums):
    prefix = [0]
    
    for x in nums:
        prefix.append(prefix[-1]+x)
        
    return prefix
```

```python
nums = [1,2,3,4,5]
```

```python
prefix = build_prefix(nums)

left = 1
right = 3

range_sum = prefix[right+1]-prefix[left]
range_sum
```

```python
# Building: O(n)

# Each range query: O(1)

# This becomes extremely useful when there are many range queries.
```
