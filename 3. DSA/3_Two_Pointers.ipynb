This is one of the most important DSA patterns.

Instead of repeatedly scanning an array, maintain two indices.

```python
nums = [1,2,3,4,5,6]

left = 0
right = len(nums)-1

while left<=right:
    print(nums[left],nums[right])
    left+=1
    right-=1
```

```python
# The pointers move toward each other.
```

```python
def reverse(nums):
    left = 0
    right = len(nums)-1
    
    while left<right:
        nums[left],nums[right] = nums[right],nums[left]
        left+=1
        right-=1
        
    return nums

reverse([1,2,3,4,5])
```

```python
# When to think "Two Pointers"

# Usually when you see:

# sorted array
# pair of elements
# palindrome
# reverse
# elements from both ends
# remove duplicates
# target sum
```
