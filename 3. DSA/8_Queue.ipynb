```python
queue = []
queue.append(10)
queue.append(20)
queue.append(30)
queue
```

```python
# This:

queue.pop(0)

# is O(n) because all remaining elements may shift.
```

```python
queue
```

```python
# Use Deque
from collections import deque
q = deque()

q.append(10)
q.append(20)
q.append(30)

print(q)
```

```python
# Remove from front:
print(q.popleft())
q
```

Complexity  
| Operation           | Complexity |
| ------------------- | ---------  |
| Enqueue `append()`  |       O(1) |
| Dequeue `popleft()` |       O(1) |
| Front               |       O(1) |
