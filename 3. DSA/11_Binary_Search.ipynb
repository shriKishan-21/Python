__Condition__  
Binary search normally requires a sorted search space.

```python
def binary_search(nums,target):
    left = 0
    right = len(nums)-1
    while right>=left:
        mid = left+(right-left)//2
        
        if nums[mid]==target:
            return mid
        
        elif nums[mid]>target:
            right=mid-1
        else:
            left=mid+1
    return -1
```

```python
binary_search([1,2,3,4,5],4)
```

__Complexity__

Time : O(log n)  
Space : O(1)

for the iterative implementation.
