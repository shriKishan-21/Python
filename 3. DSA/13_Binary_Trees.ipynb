```python
class TreeNode:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
```

```python
root = TreeNode(1)

root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

root.right.left = TreeNode(6)
root.right.right = TreeNode(7)
```

```python
# Tree Traversals

# There are four important patterns:

# DFS:
#     Preorder
#     Inorder
#     Postorder

# BFS:
#     Level order
```

```python
# Inorder
def inorder(root):
    if root is None:
        return
    
    preorder(root.left)
    print(root.data,end=" ")
    preorder(root.right)
    
# Time: O(n)
# Space: O(h)
# where h is tree height because of recursion.
```

<div class = "alert alert-warning">

__For a BST, inorder traversal produces elements in sorted order.__

```python
# Preorder
def preorder(root):
    if root is None:
        return
    
    print(root.data,end=" ")
    preorder(root.left)
    preorder(root.right)
    
# Time: O(n)
# Space: O(h)
# where h is tree height because of recursion.
```

```python
# Postorder
def postorder(root):
    if root is None:
        return
    
    preorder(root.left)
    preorder(root.right)
    print(root.data,end=" ")
    
# Time: O(n)
# Space: O(h)
# where h is tree height because of recursion.
```

<div class = "alert alert-warning">

__When useful?__

Postorder is especially useful when you need children processed before parent.  

Examples:  

deleting a tree   
calculating subtree information  
expression trees  
some dynamic programming problems  

```python
# Level Order — BFS
from collections import deque

def level_order_traversal(root):
    if root is None:
        return 
    
    queue = deque()
    queue.append(root)
    
    while queue:
        node = queue.popleft()
        
        print(node.data,end=" ")
        
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
            
# Time : O(n)
# Space : O(w)
# where w is the maximum width of the tree.
```

```python
# You're solving the problem for:

# left subtree
# +
# right subtree

# and combining their results.

# This is basically divide and conquer on a tree.
```

```python
def height(root):
    if root is None:
        return 0
    
    left = height(root.left)
    right = height(root.right)
    
    return max(left,right)+1

# Time : O(n)
# Space : O(h)
```
