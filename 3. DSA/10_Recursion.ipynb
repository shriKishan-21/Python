```python
def countdown(n):
    if n==0:
        return
    
    print(n)
    countdown(n-1)
```

```python
countdown(5)
```

<div class = "alert alert-success">
__The most important parts are:__

Base case - When recursion stops:  
if n == 0:  
    return  

Recursive case - The function calls itself:  
countdown(n - 1)  

<div class = "alert alert-success">
__Recursion and call stack__

For:

countdown(3)

think:

countdown(3)  
    ↓  
countdown(2)  
    ↓  
countdown(1)  
    ↓  
countdown(0)  

Then the calls return upward.

```python
# factorial
def factorial(n):
    if n==0 or n==1:
        return 1
    
    return n*factorial(n-1)
```

```python
factorial(5) 
```

```python
# TC : O(n)
# SC : O(n)
```
