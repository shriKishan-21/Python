### Bubble Sort  
Repeatedly compare neighboring elements and swap them when they're out of order.

```python
def bubble_sort(nums):
    n = len(nums)
    
    for i in range(n):
        for j in range(0,n-i-1):
            if nums[j]>nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
    return nums
```

```python
nums = [5,3,1,2,4]
print(bubble_sort(nums))
```

Average/worst-case time:

O(n²)

Important for learning, but generally not what you'd choose in real Python code.

### Selection Sort
Find the smallest remaining element and place it at the current position.

```python
def selection_Sort(nums):
    n = len(nums)
    for i in range(n):
        minIdx = i
        for j in range(i+1,n):
            if nums[minIdx]>nums[j]:
                minIdx=j
                
        nums[i],nums[minIdx] = nums[minIdx],nums[i]
    return nums      
```

```python
nums = [5,3,1,2,4]
print(selection_Sort(nums))
```

Time:

O(n²)

### Insertion Sort
Build the sorted portion one element at a time.

```python
def insertion_sort(nums):
    for i in range(1,len(nums)):
        key = nums[i]
        j = i-1
        while j>=0 and nums[j]>key:
            nums[j+1] = nums[j]
            j-=1
        nums[j+1] = key
        
    return nums
```

```python
nums = [5,3,1,2,4]
print(insertion_sort(nums))
```

Worst case:

O(n²)

But it performs well on small or nearly sorted data.

### Merge Sort
Merge sort uses divide and conquer.

```python
def merge(nums,s,e):
    mid = (s+e)//2
    
    idx = mid+1
    
    if nums[mid] <= nums[idx]:
        return
    
    while s<=mid and idx<=e:
        if nums[s]<=nums[idx]:
            s+=1
        else:
            value = nums[idx]
            index = idx
            while index!=s:
                nums[index] = nums[index-1]
                index-=1
            nums[s]=value
            s+=1
        mid+=1
        idx+=1
    
    
    
def merge_sort(nums,start,end):
    if start>=end:
        return
     
    mid = (start+end)//2
    merge_sort(nums,start,mid)
    merge_sort(nums,mid,end)
    merge(nums,start,end)
    return nums    
    
```

```python
nums = [8,5,7,2,3,6,1,4,9]
print(merge_sort(nums,0,len(nums)))
```

Time: O(n log n)  
Extra space: O(n)

### Python's Real-World Sorting

```python
# In actual Python code, you'll usually use:

# nums.sort()
# or:
# sorted(nums)
```

```python
# Example:

nums = [5, 2, 8, 1]

result = sorted(nums)

print(result)
print(nums)

# Output:
# [1, 2, 5, 8]
# [5, 2, 8, 1]
```

<div class="alert">
Difference:  
  
list.sort()   → modifies the existing list  
sorted()      → returns a new sorted object  

Python's built-in sorting is highly optimized and based on Timsort.
