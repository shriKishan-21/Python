### __Frequency counting__
One of the most common DSA patterns:

```python
nums = [1,2,2,3,3,3]

freq = {}

for x in nums:
    freq[x] = freq.get(x,0)+1
    
print(freq)
```

```python
# Python also provides:

from collections import Counter

freq = Counter(nums)
freq
```
