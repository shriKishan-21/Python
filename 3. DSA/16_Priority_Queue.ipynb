```python
# A priority queue means:
# Elements are removed according to priority, not insertion order.
```

```python
# Heap is the normal implementation.
```

```python
import heapq

tasks = []

heapq.heappush(tasks,(2,"Low Priority"))
heapq.heappush(tasks,(1,"High Priority"))
heapq.heappush(tasks,(3,"Very Low Priority"))


print(heapq.heappop(tasks))
```

Smaller number = higher priority in this example.

Priority queues are heavily used in:  
scheduling  
Dijkstra  
top-k problems  
event simulation  
merging sorted streams
