```python
# Sliding window is basically a controlled moving range.
```

```python
# Suppose:

# [2, 1, 5, 1, 3, 2]

# We want the maximum sum of 3 consecutive elements.

# A brute-force approach might repeatedly calculate:

# 2 + 1 + 5
# 1 + 5 + 1
# 5 + 1 + 3
# 1 + 3 + 2
```

```python
# Instead, use a window.

# Start:

# 2 1 5 | 1 3 2
# ─────
#   8

# Move right:

# Remove 2, add 1.

# 2 | 1 5 1 | 3 2
#     ─────
#       7

# Then:

# 2 1 | 5 1 3 | 2
#        ─────
#          9

# Then:

# 2 1 5 | 1 3 2
#           ─────
#             6

# Maximum = 9.
```

```python
def max_sum(nums,k):
    # initial window sum - sum of first k index
    window_sum = sum(nums[:k])
    # store maximum sum in window
    maximum = window_sum
    
    for right in range(k,len(nums)):
        window_sum += nums[right]
        window_sum -= nums[right-k]
        
        maximum = max(maximum,window_sum)
        
    return maximum
```

```python
max_sum([1,2,3,4,5,0,3,7],2)
```

```python
# Time: O(n)

# Without sliding window, a naive approach can be O(nk).
```
