| Operation      |     Complexity |
| -------------- | -------------: |
| `nums[i]`      |           O(1) |
| `nums[i] = x`  |           O(1) |
| `append()`     | O(1) amortized |
| `pop()`        |           O(1) |
| `insert(0, x)` |           O(n) |
| `pop(0)`       |           O(n) |
| Search         |           O(n) |

```python
# Different methods to create array in pythons
# 1. Using [] — most common
arr = [1, 2, 3, 4, 5]

print(arr)
```

```python
# 2 Empty array/list
arr = []
arr.append(10)
arr.append(20)
print(arr)
```

```python
# 3. using list()
arr = list()
# or
arr = list((1,2,3,4,5))
print(arr)
```

```python
# 4. Using range()
arr = list(range(5))

print(arr)
```

```python
# 5. List comprehension
arr = [i for i in range(5)]
print(arr)
```

```python
# Create an array of fixed size
arr = [0]*5
print(arr)
```

```python
# 7. Taking input into an array
arr = list(map(int,input().split()))
print(arr)
```

```python
# 8. using array module
from array import array
arr = array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
# 9. NumPy array
import numpy as np
arr = np.array([1,2,3,4,5])
print(arr)
```
